```python
import gzip, json
import numpy as np
import pandas as pd

rows = []
for p in sorted(root.glob(f"*{pdb}*ipca*.json.gz")):
    with gzip.open(p, "rt", encoding="utf-8") as f:
        d = json.load(f)
    V = np.asarray(d["components_"], dtype=float)
    evr = np.asarray(d.get("explained_variance_ratio_", []), dtype=float)

    rows.append({
        "file": p.name,
        "n_pc": V.shape[0],
        "n_feat": V.shape[1],
        "xyz_stacked": (V.shape[1] % 3 == 0),
        "evr_sum": float(evr.sum()) if evr.size else np.nan,
        "evr1": float(evr[0]) if evr.size else np.nan,
        "mtime": p.stat().st_mtime,
    })

df = pd.DataFrame(rows).sort_values(["xyz_stacked","mtime"], ascending=[False, False])
df
```
