```python
import mdtraj as md

PDBID = "7lak"
replica = 5
PC_num = 1
NMs = 1000
NMs2 = 10

pdb1 = f"results/{PDBID}/replica_{replica}/pc1_nma_projection/rep{replica}_pc{PC_num}_lowest{NMs}nm_minus.pdb"
#pdb2 = f"results/{PDBID}/replica_0/pc1_nma_projection/stopped.pdb"
pdb2 = f"results/{PDBID}/replica_{replica}/pc1_nma_projection/rep{replica}_pc{PC_num}_lowest{NMs2}nm_minus.pdb"

ref = md.load(pdb2)
mob = md.load(pdb1)

# Use C-alpha atoms for alignment and RMSD
ref_sel = ref.topology.select("protein and name CA")
mob_sel = mob.topology.select("protein and name CA")

if len(ref_sel) != len(mob_sel):
    raise ValueError(
        f"Different number of selected atoms: "
        f"ref={len(ref_sel)}, mobile={len(mob_sel)}"
    )

# Align mobile onto reference using the same atoms
mob.superpose(
    ref,
    atom_indices=mob_sel,
    ref_atom_indices=ref_sel,
)

# Calculate RMSD after alignment
rmsd_nm = md.rmsd(
    mob,
    ref,
    atom_indices=mob_sel,
)

mob.save_pdb(f"results/{PDBID}/replica_{replica}/pc1_nma_projection/{PDBID}_rep{replica}_aligned{NMs}_{NMs2}.pdb")
print(f"Aligned structure saved to results/{PDBID}/replica_{replica}/pc1_nma_projection/aligned{NMs}_{NMs2}.pdb")

print(f"Aligned C-alpha RMSD: {rmsd_nm[0]:.4f} nm")
print(f"Aligned C-alpha RMSD: {rmsd_nm[0] * 10:.4f} Å")
```
